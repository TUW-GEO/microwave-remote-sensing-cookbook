```python
import ipywidgets as widgets
```

```python
def create_checkbox(answer: list[str]) -> list[widgets.Checkbox]:
    """Test."""
    cb_list = []
    for a in answer:
        cb = widgets.Checkbox(value=False, description=a, disabled=False, indent=False)
        cb_list.append(cb)

    return cb_list


def show_checkbox(cb: list[widgets.Checkbox]) -> None:
    """Test."""
    box = widgets.VBox(cb)
    display(box)
```
